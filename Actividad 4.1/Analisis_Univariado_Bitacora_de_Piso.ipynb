# Análisis univariado - Bitácora de Piso

```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde el archivo .csv ya limpio (ajusta la ruta si no esta en la misma carpeta que este notebook)
Bitacora_Piso = pd.read_csv('Bitacora_de_Piso_Limpia.csv', encoding='utf-8-sig')
```

```python
#Verificamos información del DataFrame
Bitacora_Piso.info()
```

```python
#Imprimo los primeros 5 registros del dataframe
Bitacora_Piso.head()
```

## Análisis univariado: Venta

```python
#Obtengo un análisis univariado de las variables categóricas
Tabla_freq = Bitacora_Piso['Venta'].value_counts().reset_index()
Tabla_freq
```

```python
#Obtengo un filtro de los valores más reelevantes de la variable categórica seleccionada
Filtro= Tabla_freq[Tabla_freq['count']>1]
Filtro
```

```python
#Ajusto el indice de mi dataframe
Filtro_index= Filtro.set_index('Venta')
Filtro_index
```

```python
#Realizamos grafico de barras del dataframe filtrado
Filtro_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= "red")
plt.title('Ventas concretadas')
plt.xlabel('Venta')
plt.ylabel('Frecuencia')
```

```python
#Realizamos grafico de barras del dataframe filtrado
Filtro_index.plot(kind = 'bar', width=0.8, figsize=(10,4), color= "red")
plt.title('Ventas concretadas')
plt.xlabel('Venta')
plt.ylabel('Frecuencia')
```

```python
#Realizamos grafico de pastel del dataframe filtrado
Filtro_index["count"].plot(kind='pie', figsize=(10,5), shadow=True, autopct="%0.1f %%")
plt.title('Ventas concretadas')
```
