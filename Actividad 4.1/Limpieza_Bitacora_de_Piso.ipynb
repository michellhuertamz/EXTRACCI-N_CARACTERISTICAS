# Limpieza de la Bitácora de Piso

```python
import pandas as pd

# Ajusta esta ruta si el archivo no esta en la misma carpeta que este notebook
archivo_entrada = r"C:\Users\yarem\Desktop\Actividad_4.1\Analisis GAC Full7agosto 2026(Bitacora de Piso).csv"

# encoding="utf-8-sig" quita el BOM del archivo (evita que la primera columna se lea como "Fecha" con caracteres raros al inicio)
df = pd.read_csv(archivo_entrada, encoding="utf-8-sig")

print("Forma original:", df.shape)
df.head()
```

## 2. Eliminar columnas

Se eliminan `Teléfono` y `Estatus de SDC` 

```python
# quitamos espacios sobrantes en los nombres de columna (ej. "Telefono " -> "Telefono")
df.columns = df.columns.str.strip()

# las columnas "Unnamed" son las columnas sin encabezado real que traen "Items"
columnas_items = [col for col in df.columns if col.startswith("Unnamed")]

columnas_a_eliminar = ["Telefono", "Estatus de SDC"] + columnas_items
df = df.drop(columns=columnas_a_eliminar)

print("Columnas eliminadas:", columnas_a_eliminar)
print("Forma tras eliminar columnas:", df.shape)
```

```python
filas_57_a_67 = range(55, 66)      # fila 57 a 67 de tu hoja
filas_210_a_235 = range(208, 234)  # fila 210 a 235 de tu hoja

df = df.drop(index=list(filas_57_a_67) + list(filas_210_a_235))
df = df.reset_index(drop=True)

print("Forma tras eliminar filas:", df.shape)
print("Filas duplicadas restantes:", df.duplicated().sum())
```

```python
for columna in df.columns:
    df[columna] = df[columna].apply(lambda valor: valor.strip() if isinstance(valor, str) else valor)

# normalizar Fecha a mm/dd/aaaa siempre con 2 digitos
df["Fecha"] = pd.to_datetime(df["Fecha"], format="%m/%d/%Y", errors="coerce").dt.strftime("%m/%d/%Y")

print("Ejemplo de fechas:", df["Fecha"].head(5).tolist())
```

```python
df["Esatus Lead"] = df["Esatus Lead"].replace({"venta": "Venta"})
df["Temperatura"] = df["Temperatura"].replace({"Lead InteresBajo": "Lead Interes Bajo"})

print("Esatus Lead:", sorted(df["Esatus Lead"].dropna().unique()))
print("Temperatura:", sorted(df["Temperatura"].dropna().unique()))
```

```python
nulos_antes = int(df.isna().sum().sum())
df = df.fillna("Sin dato")

print(f"Se rellenaron {nulos_antes} valores faltantes con 'Sin dato'")
```

```python
print("Forma final:", df.shape)
print("Valores nulos restantes:", df.isna().sum().sum())
print("Filas duplicadas restantes:", df.duplicated().sum())

archivo_salida = "Bitacora_de_Piso_Limpia.csv"
df.to_csv(archivo_salida, index=False, encoding="utf-8-sig")
print("Archivo guardado como:", archivo_salida)

df.head()
```
